```python
import torch
import cv2
import numpy as np
```

```python
import pathlib
from pathlib import Path
pathlib.PosixPath = pathlib.WindowsPath
```

```python
model = torch.hub.load('ultralytics/yolov5', 'custom', path='xxxxxxxxxxxxxxxxxxxxxxxxxx/xxx.pt')
```

```python
img_path = 'xxxxxxxxxxxxxxxxxxxxxxxxxx/xxxx.jpg'  # <- aquí pones tu imagen
img = cv2.imread(img_path)
```

```python
results = model(img)
```

```python
cv2.imshow("Anomaly", np.squeeze(results.render()))
cv2.waitKey(0)
cv2.destroyAllWindows()
```
