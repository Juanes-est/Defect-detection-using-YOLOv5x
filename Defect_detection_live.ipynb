```python
import torch
import cv2
import numpy as np
```

```python
import pathlib
from pathlib import Path
pathlib.PosixPath = pathlib.WindowsPath
```

```python
model = torch.hub.load('ultralytics/yolov5', 'custom', path='xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx/xxxx.pt')
```

```python
cap = cv2.VideoCapture(0)


while True:
    # frame reading
    ret, frame = cap.read()

    # color correction
    #frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # execution of the detection process
    detect = model(frame)

    info = detect.pandas().xyxy[0]  # im1 predictions
    print(info)

    # FPS
    cv2.imshow('Defect Detection', np.squeeze(detect.render()))

    t = cv2.waitKey(5)
    if t == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
